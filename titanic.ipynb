```python
import pandas as pd
df = pd.read_csv('titanic.csv')
df.head()
#df.shape
```

```python
#PROBLEMS:
# 1.Total people travelling
df['PassengerId'].count()
```

```python
# 2.How many survived/died
df['Survived'].sum()
df[df['Survived'] == 0]['PassengerId'].count()
```

```python
# 3.a)How many elders were there?
df[df['Parch'] != 0]['Parch'].sum()

```

```python
# 3. b) How many elders survived?
df['Elders Survived'] = df['Parch'] * df['Survived']
```

```python
# 3. b) How many elders survived?
(df['Elders Survived'].sum()/df[df['Parch'] != 0]['Parch'].sum())*100
```

```python
# 4.How many kids were there?
df[(df['Age'] < 18)]['PassengerId'].count()
```

```python
#4.b)How many survived?
df[(df['Age'] < 18) & (df['Survived'] != 0)]['PassengerId'].count()
```

```python
# 5. How many females were there?How many survived?
df[df['Sex'] == 'female'].count()
df[(df['Sex'] == 'female') & df['Survived'] != 0]['Survived'].sum()
```

```python
# 6.How many were rich and poor were there?How many survived?
# If fare is upto 15, call it 'lower class'
# If fare is between 15 - 32, call it 'middle class'
# If fare is between 32 - 100, call it 'upper class'
# If fare is more than 100, call it 'elites'
df['Economic classes']=df.apply(lambda row:'lower class' if row['Fare'] <= 15 else 'middle class' if row['Fare']>=15 and row['Fare'] <= 32 else 'upper class' if row['Fare'] > 32 and row['Fare'] <= 100 else 'elites',axis=1)
df.groupby('Economic classes')['Survived'].count()
df.groupby('Economic classes')['Survived'].sum()

```

```python
# 7.What's survival ratio of people boarding on each port?
(df[df['Survived'] == 1].groupby('Embarked')['Survived'].count()/df.shape[0])*100
```

```python
# 8.How many were travelling solo? What ratio of those survived? 
(df[(df['SibSp'] == 0) & (df['Parch'] == 0)]['Survived'].sum()/(df[(df['SibSp'] == 0) & (df['Parch'] == 0)].shape[0]))*100
```

```python
# 9.a)How many were not travelling solo?
df_non_solo_travellers = df[(df['SibSp'] != 0)|(df['Parch'] != 0)]
```

```python
# 9.a)How many were not travelling solo?
len(df_non_solo_travellers)
```

```python
#9. b)what ratio of those survived?
(df_non_solo_travellers['Survived'].sum()/(df_non_solo_travellers.shape[0]))*100
```
